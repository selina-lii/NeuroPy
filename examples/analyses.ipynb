Let's point our data residing inside `basepath` folder to `processData`

```python
%matplotlib inline
from callfunc import processData
from plotUtil import Fig
basepath = "/data/Clustering/SleepDeprivation/RatS/Day3SD/"
sess = processData(basepath)
```

Create track epochs from existing epochs and plot those

```python
sess.tracks.create(epoch_names=['maze1','maze2'])
sess.tracks.plot()

```

These are circular so let's linearize the coordinates

```python
sess.tracks.linearize_position(track_name=['maze1','maze2'])
```

Calculating run epochs for `maze1` epoch. You can adjust various parameters to your needs but here everything is kept at default values. Green epochs in the figure show forward running and red epochs show running in opposite direction. 


```python
sess.tracks.estimate_run_laps('maze1')
```

Let's compute place fields using these run laps

```python
sess.placefield.pf1d.compute('maze1',run_dir='forward')
sess.placefield.pf1d.plot(normalize=True)
```
